# Preprocessing

## Load medical data and patient reported data and format ID and columns properly

```python
extracted_medical_data = pd.read_excel("received_tables/Batches_1-12_Dates_Notes_Masked (4).xlsx")
patient_reported_data = pd.read_excel("received_tables/ASC PRD Cleaned_200 pts.xlsx")

extracted_medical_data["STUDY ID"] = extracted_medical_data["STUDY ID"].apply(lambda x: x.replace("ASCPROJECT","ASCProject"))
extracted_medical_data.columns = [col.strip() for col in extracted_medical_data.columns]
extracted_medical_data["PRIMARY SITE"] = extracted_medical_data["PRIMARY SITE"].str.strip()
extracted_medical_data["PRIMARY SITE STANDARDIZED"] = extracted_medical_data["PRIMARY SITE"].replace({
    #"BREAST (PARENCHYMAL)": "BREAST",
    #"BREAST (CUTANEOUS)": "BREAST",
    "TRUNK": "OTHER",
    "NOT FOUND IN RECORD": "NA"
})

extracted_medical_data.loc[
    extracted_medical_data["DX BIOPSY LOCATION"].str.contains("LUNG") & (extracted_medical_data["PRIMARY SITE"] == "VISCERAL ORGAN"),
    "PRIMARY SITE STANDARDIZED"
] = "LUNG"

#print(extracted_medical_data["PRIMARY SITE STANDARDIZED"].value_counts())

extracted_medical_data.loc[
    extracted_medical_data["DX BIOPSY LOCATION"].str.contains("LIVER") & (extracted_medical_data["PRIMARY SITE"] == "VISCERAL ORGAN"),
    "PRIMARY SITE STANDARDIZED"
] = "LIVER"

#print(extracted_medical_data["PRIMARY SITE STANDARDIZED"].value_counts())


extracted_medical_data.loc[
    extracted_medical_data["PRIMARY SITE"].str.contains("BONE"),
    "PRIMARY SITE STANDARDIZED"
] = "BONE"

#print(extracted_medical_data["PRIMARY SITE STANDARDIZED"].value_counts())


extracted_medical_data.loc[
    extracted_medical_data["DX BIOPSY LOCATION"].str.contains("ARM|LEG"),
    "PRIMARY SITE STANDARDIZED"
] = "EXTREMITY"

extracted_medical_data.loc[
    extracted_medical_data["DX BIOPSY LOCATION"].str.contains("CHEST"),
    "PRIMARY SITE STANDARDIZED"
] = "BREAST (CUTANEOUS)"

extracted_medical_data.loc[
    extracted_medical_data["DX BIOPSY LOCATION"].str.contains("BACK"),
    "PRIMARY SITE STANDARDIZED"
] = "OTHER"

extracted_medical_data.loc[
    extracted_medical_data["PRIMARY SITE STANDARDIZED"].str.contains("OTHER"),
    "PRIMARY SITE STANDARDIZED"
] = "OTHER"

extracted_medical_data["RAAS"] = extracted_medical_data["RAAS"].replace({"YES":1,"NO":0,"NOT FOUND IN RECORD":np.NaN})
extracted_medical_data["LAAS"] = extracted_medical_data["LAAS/STEWART-TREVES"].replace({"YES":1,"NO":0,"NOT FOUND IN RECORD":np.NaN})
extracted_medical_data["CUTANEOUS AS"] = extracted_medical_data["CUTANEOUS AS"].replace({"YES":1,"NO":0,"NOT FOUND IN RECORD":np.NaN})
extracted_medical_data["LOCAL RECURRENCE"] = extracted_medical_data["LOCAL RECURRENCE"].replace({"YES":1,"NO":0,"NOT FOUND IN RECORD":np.NaN})
extracted_medical_data["HAS METS AT DX"] = extracted_medical_data["METS SITES AT METS DX"].replace({
    "NOT METS": "NO METS",
})

extracted_medical_data.loc[
    extracted_medical_data["HAS METS AT DX"] != "NO METS","HAS METS AT DX"
] = 1
extracted_medical_data.loc[
    extracted_medical_data["HAS METS AT DX"] == "NO METS","HAS METS AT DX"
] = 0
extracted_medical_data.loc[
    extracted_medical_data["METS SITES AT METS DX"] == "NOT FOUND IN RECORD","HAS METS AT DX"
] = np.NaN

extracted_medical_rename_dict = {
    col:col + " (EHR_EXTRACTED)" for col in extracted_medical_data.columns if col != "STUDY ID"
}
extracted_medical_data = extracted_medical_data.rename(columns=extracted_medical_rename_dict)

def format_asc_id(string):
    return "ASCProject_" + str(string).zfill(4)

patient_reported_data["STUDY ID"] = patient_reported_data["legacy_shortid"].apply(format_asc_id)
patient_reported_data_rename_dict = {
    col:col + " (PRD)" for col in patient_reported_data.columns if col != "STUDY ID"
}
patient_reported_data = patient_reported_data.rename(columns=patient_reported_data_rename_dict)

```

```python
extracted_medical_data["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"].value_counts()
```

## Use MED EXTRACTED whenever possible and PRD if not

```python
merged_clin_data = extracted_medical_data.merge(patient_reported_data,on="STUDY ID",how="outer")
merged_clin_data["EHR_EXTRACTED"] = merged_clin_data["STUDY ID"].isin(extracted_medical_data["STUDY ID"])

## Combine Age
merged_clin_data["Age (Combined)"] = merged_clin_data["AGE AT DX (EHR_EXTRACTED)"].fillna(
    merged_clin_data["AGE OF DIAGNOSIS (PRD)"]
).replace("NOT FOUND IN RECORD",np.NaN).astype(float)

## Combine Primary Sites
merged_clin_data["PRIMARY SITE (Combined)"] = merged_clin_data["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"].fillna(
    merged_clin_data["DIAGNOSIS_LOC (PRD)"]
).replace("NOT FOUND IN RECORD",np.NaN)

multiple_primaries_combined_index = merged_clin_data["PRIMARY SITE (Combined)"].apply(lambda x: len(x.split(",")) > 1)

merged_clin_data.loc[
    multiple_primaries_combined_index,
    "PRIMARY SITE (Combined)"
] = np.NaN

merged_clin_data.loc[
    merged_clin_data["PRIMARY SITE (Combined)"] == "NA",
    "PRIMARY SITE (Combined)"
] = np.NaN

merged_clin_data.loc[
    (merged_clin_data["PRIMARY SITE (Combined)"] == "BREAST"),
    "PRIMARY SITE (Combined)"
] = "BREAST (NOS)"

merged_clin_data.loc[
    (merged_clin_data["PRIMARY SITE (Combined)"] == "ABDOMINAL"),
    "PRIMARY SITE (Combined)"
] = "VISCERAL ORGAN"

merged_clin_data.loc[
    (merged_clin_data["PRIMARY SITE (Combined)"] == "BONE") | (merged_clin_data["PRIMARY SITE (Combined)"] == "EXTREMITY"),
    "PRIMARY SITE (Combined)"
] = "BONELIMB"

merged_clin_data.loc[
    (merged_clin_data["PRIMARY SITE (Combined)"] == "HEADFACENECK") | (merged_clin_data["PRIMARY SITE (Combined)"] == "SCALP"),
    "PRIMARY SITE (Combined)"
] = "HNFS"

merged_clin_data.loc[
    (merged_clin_data["PRIMARY SITE (Combined)"] == "CARDIAC"),
    "PRIMARY SITE (Combined)"
] = "HEART"


merged_clin_data["PRIMARY SITE (Combined)"].value_counts()
```

## Make RAAS/LAAS classifications

```python
def make_raas_laas_classification(row):
    if (row["LAAS (EHR_EXTRACTED)"] != row["LAAS (EHR_EXTRACTED)"]) and (row["RAAS (EHR_EXTRACTED)"] != row["RAAS (EHR_EXTRACTED)"]):
        return np.NaN
    laas_flag = row["LAAS (EHR_EXTRACTED)"] == 1
    raas_flag = row["RAAS (EHR_EXTRACTED)"] == 1
    if raas_flag and laas_flag:
        return "RAAS & LAAS"
    elif laas_flag:
        return "LAAS"
    elif raas_flag:
        return "RAAS"
    else:
        return "Non-RAAS/LAAS"

raas_laas_classes = [make_raas_laas_classification(row) for idx,row in merged_clin_data.iterrows()]
merged_clin_data["RAAS_LAAS_Class"] = raas_laas_classes
merged_clin_data

```

## What are the most common metastatic events?

```python
def split_string(input_string):
    words = []
    bracket_count = 0
    current_word = ''
    for char in input_string:
        if char == ',' and bracket_count == 0:
            words.append(current_word.strip())
            current_word = ''
        else:
            current_word += char
        if char == '[':
            bracket_count += 1
        elif char == ']':
            bracket_count -= 1
    words.append(current_word.strip())
    return words

def format_met_sites(string):
    if string != string:
        return np.NaN
    else:
        ## Replace () with []
        string = string.replace("(","[").replace(")","]")
        splitted_string = split_string(string)
        return splitted_string
        

metastatic_event_table_rows = []
for idx,row in med_clin_data.iterrows():
    patient_id = row["STUDY ID"]
    primary_site = row["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"]
    met_sites = row["METS SITES EVER (EHR_EXTRACTED)"]
    
    if met_sites != met_sites:
        metastatic_event_table_rows.append({
            "STUDY ID": patient_id,
            "PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)": primary_site,
            "MET SITE": np.NaN
        })
    else:
        for site in format_met_sites(met_sites):
            metastatic_event_table_rows.append({
                "STUDY ID": patient_id,
                "PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)": primary_site,
                "MET SITE": site.strip()
            })

metastatic_event_df = pd.DataFrame(data=metastatic_event_table_rows)
metastatic_event_df = metastatic_event_df.replace("NOT FOUND IN RECORD",np.NaN).replace("NA",np.NaN)
metastatic_event_df = metastatic_event_df.dropna()
metastatic_event_df
```

```python
## Do some formatting
metastatic_event_df["MET SITE STANDARDIZED"] = metastatic_event_df["MET SITE"]
metastatic_event_df.loc[
    metastatic_event_df["MET SITE"].str.contains("OTHER"),
    "MET SITE STANDARDIZED"
] = "OTHER"

## For simplicitiy, met sites below top 10 frequencies are lumped into OTHER
below_top_10_met_sites = metastatic_event_df["MET SITE STANDARDIZED"].value_counts()[9:].index.tolist()
metastatic_event_df.loc[
    metastatic_event_df["MET SITE STANDARDIZED"].isin(below_top_10_met_sites),
    "MET SITE STANDARDIZED"
] = "OTHER"
## Avoid counting "OTHER" twice per sample
metastatic_event_df = metastatic_event_df.drop_duplicates(subset=["STUDY ID","MET SITE STANDARDIZED"]).copy()

metastatic_event_df.loc[
    metastatic_event_df["MET SITE"].str.contains("DISTANT LYMPH NODE"),
    "MET SITE STANDARDIZED"
] = "DISTANT LYMPH NODE"
metastatic_event_df["MET SITE STANDARDIZED"].value_counts()
```

```python
metastatic_event_ct = pd.crosstab(
    metastatic_event_df["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"],
    metastatic_event_df["MET SITE STANDARDIZED"]
)
metastatic_event_ct["sum"] = metastatic_event_ct.sum(axis=1)
metastatic_event_ct = metastatic_event_ct.sort_values("sum",ascending=False).drop(columns="sum")

metastatic_event_ct_reversed = metastatic_event_ct.T
metastatic_event_ct_reversed["sum"] = metastatic_event_ct_reversed.sum(axis=1)
metastatic_event_ct_reversed = metastatic_event_ct_reversed.sort_values("sum",ascending=False).drop(columns="sum")
metastatic_event_ct_reversed = metastatic_event_ct_reversed.drop("NO METS")

metastatic_event_ct = metastatic_event_ct.rename({"BREAST (PARENCHYMAL)":"BREAST\n(PARENCHYMAL)"})
metastatic_event_ct = metastatic_event_ct.rename({"BREAST (CUTANEOUS)":"BREAST\n(CUTANEOUS)"})

metastatic_event_ct
```

## What are the rates of metastasis by primay sites? 

```python
med_clin_data.loc[
    med_clin_data["NUM MET SITES EVER (EHR_EXTRACTED)"] > 0,
    "HAS_MET",
] = True
med_clin_data.loc[
    med_clin_data["NUM MET SITES EVER (EHR_EXTRACTED)"] == 0,
    "HAS_MET",
] = False

met_rate_by_site_ct = pd.crosstab(med_clin_data["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"],med_clin_data["HAS_MET"])
met_rate_by_site_ct["sum"] = met_rate_by_site_ct.sum(axis=1)
met_rate_by_site_ct["rate"] = met_rate_by_site_ct[True]/met_rate_by_site_ct["sum"]
met_rate_by_site_ct = met_rate_by_site_ct.drop("NA")
met_rate_by_site_ct = met_rate_by_site_ct.sort_values("rate",ascending=False)
met_rate_by_site_ct
```

# What are the most common treatments?

```python
treatment_ehr_df = pd.read_excel("received_tables/Batches_1-12_Dates_Notes_Masked (4).xlsx",sheet_name="TX")
treatment_ehr_df["STUDY ID"] = treatment_ehr_df["STUDY ID"].apply(lambda x: x.replace("ASCPROJECT","ASCProject"))
treatment_ehr_df.columns = [col.strip() for col in treatment_ehr_df.columns]
treatment_ehr_df = treatment_ehr_df.replace({"NOT FOUND IN RECORD":np.NaN,"NA": np.NaN})
treatment_ehr_df = treatment_ehr_df.dropna(subset=["DRUG","MODE"])
treatment_ehr_df["DRUG"] = treatment_ehr_df["DRUG"].str.strip()
treatment_ehr_df = treatment_ehr_df.sort_values(by=["STUDY ID","START DATE (DAYS FROM DX)"])
treatment_ehr_df = treatment_ehr_df[~treatment_ehr_df["MODE"].str.contains("OTHER")]
treatment_ehr_df_merged = treatment_ehr_df.merge(merged_clin_data,on="STUDY ID",how="left")
treatment_ehr_df_merged
```

```python
tx_one_per_patient = treatment_ehr_df_merged.drop_duplicates(subset=["STUDY ID","DRUG"]).copy()
drug_counts = tx_one_per_patient["DRUG"].value_counts()
drugs_to_omit = drug_counts[9:].index.tolist()
tx_one_per_patient.loc[
    tx_one_per_patient["DRUG"].isin(drugs_to_omit),
    "DRUG"
] = "OTHER"
tx_one_per_patient.loc[
    tx_one_per_patient["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"].str.contains("BREAST"),
    "PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"
] = "BREAST" 
tx_one_per_patient
```

```python
tx_one_per_patient_ct = pd.crosstab(
    tx_one_per_patient["DRUG"],
    tx_one_per_patient["PRIMARY SITE STANDARDIZED (EHR_EXTRACTED)"]
)
tx_one_per_patient_ct["sum"] = tx_one_per_patient_ct.sum(axis=1)
tx_one_per_patient_ct = tx_one_per_patient_ct.drop("OTHER")
tx_one_per_patient_ct = tx_one_per_patient_ct.drop(columns="NA")

tx_one_per_patient_ct = tx_one_per_patient_ct.sort_values("sum",ascending=False).drop(columns="sum")
tx_one_per_patient_ct
```
